```python
marks=[78,85,90,66,88]
print(max(marks)) # Maximum marks
print(min(marks)) # Minimum marks
print(sum(marks)) # Total of marks
print(len(marks)) # Length of marks
print(sum(marks)/len(marks)) # Average of marks
marks.sort() # Sort in ascending order
print(marks)
```
